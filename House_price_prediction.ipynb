```python
#python
# ============================================
# MY SECOND MACHINE LEARNING PROJECT
# Goal: Teach a computer to guess a house's price
#       based on things like its size and location.
# ============================================

# Step 1: Get the housing data
# sklearn has a real California housing dataset built in — no download needed.
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()

# The features for each house (income in area, house age, rooms, location, etc.)
house_features = housing.data

# The correct answer for each house: its price (in $100,000s)
house_prices = housing.target

print("Number of houses in dataset:", len(house_features))
print("Feature names:", housing.feature_names)
print("Example house features:", house_features[0])
print("Its actual price (in $100,000s):", house_prices[0])


# Step 2: Split the data into two groups
# - Training data: the computer studies these to learn the pattern
# - Test data: we hide these, then quiz it later
from sklearn.model_selection import train_test_split

train_features, test_features, train_prices, test_prices = train_test_split(
    house_features, house_prices, test_size=0.2
)
# test_size=0.2 means: use 80% of houses to study, 20% to quiz

print("\nHouses used for training:", len(train_features))
print("Houses used for testing (quiz):", len(test_features))


# Step 3: Create and train the model
# Linear Regression = the simplest way to predict a NUMBER (not a category).
# It finds the best straight-line relationship between features and price.
from sklearn.linear_model import LinearRegression
model = LinearRegression()

# .fit() is where the learning happens — we show it features AND correct
# prices together, so it can find the pattern connecting them.
model.fit(train_features, train_prices)

print("\nModel has finished training!")


# Step 4: Quiz the model on houses it has NEVER seen before
my_guesses = model.predict(test_features)

print("\nModel's price guesses (first 5):  ", my_guesses[:5].round(2))
print("Actual correct prices (first 5):  ", test_prices[:5].round(2))


# Step 5: Check how close the guesses were
# Since this is predicting a NUMBER (not a category like flower species),
# we can't use "accuracy" — instead we measure the average error.
from sklearn.metrics import mean_absolute_error
error = mean_absolute_error(test_prices, my_guesses)

print(f"\nOn average, the model's guess was off by ${error * 100000:,.0f}")
```
